**PROJECT 1 - STUDENT MANAGEMENT SYSTEM**

```python
from abc import ABC, abstractmethod


# ============================================================
# 1. ABSTRACT BASE CLASS
# ============================================================

class Person(ABC):

    def __init__(self, name):
        self.__name = ""
        self.set_name(name)  # uses the setter so the name gets validated

    # Getter
    def get_name(self):
        return self.__name

    # Setter
    def set_name(self, name):
        name = name.strip()
        if name == "" or "|" in name:
            raise ValueError("Name cannot be empty or contain '|'.")
        self.__name = name

    @abstractmethod
    def display_info(self):
        pass


# ============================================================
# 2. STUDENT CLASS - INHERITANCE
# ============================================================

class Student(Person):

    def __init__(self, roll_no, name, marks):
        super().__init__(name)

        # Private attributes - Encapsulation
        self.__roll_no = ""
        self.__marks = 0

        # Setters are used so values are always validated
        self.set_roll_no(roll_no)
        self.set_marks(marks)

    # Getter for Roll Number
    def get_roll_no(self):
        return self.__roll_no

    # Setter for Roll Number
    def set_roll_no(self, roll_no):
        roll_no = roll_no.strip()
        if roll_no == "" or "|" in roll_no:
            raise ValueError("Roll number cannot be empty or contain '|'.")
        self.__roll_no = roll_no

    # Getter for Marks
    def get_marks(self):
        return self.__marks

    # Setter for Marks
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            raise ValueError("Marks must be between 0 and 100.")

    # Display student information
    def display_info(self):
        print("Roll Number :", self.__roll_no)
        print("Name        :", self.get_name())
        print("Marks       :", self.__marks)
        print("----------------------------")

    # Convert student data into text format
    def to_file_format(self):
        return f"{self.__roll_no}|{self.get_name()}|{self.__marks}"


# ============================================================
# 3. FILE MANAGER CLASS - FILE HANDLING
# ============================================================

class FileManager:

    FILE_NAME = "students.txt"

    # Save students to TXT file
    def save_students(self, students):

        try:
            with open(self.FILE_NAME, "w") as file:

                for student in students:
                    file.write(student.to_file_format() + "\n")

            print("Records saved successfully!")

        except IOError:
            print("Error: Unable to save records.")

    # Load students from TXT file
    def load_students(self):

        students = []

        try:
            with open(self.FILE_NAME, "r") as file:

                for line in file:

                    # Skip empty lines
                    if line.strip() == "":
                        continue

                    # try/except is inside the loop, so one bad line
                    # is skipped and the remaining lines still load
                    try:
                        data = line.strip().split("|")

                        if len(data) != 3:
                            raise ValueError("Wrong number of fields.")

                        roll_no = data[0]
                        name = data[1]
                        marks = float(data[2])

                        student = Student(
                            roll_no,
                            name,
                            marks
                        )

                        students.append(student)

                    except ValueError:
                        print("Skipped invalid line:", line.strip())

        except FileNotFoundError:
            # File will be created when records are first saved
            pass

        except IOError:
            print("Error while reading student records.")

        return students


# ============================================================
# 4. STUDENT MANAGER CLASS
# ============================================================

class StudentManager:

    def __init__(self):
        self.students = []
        self.file_manager = FileManager()

        # Load existing records
        self.students = self.file_manager.load_students()

    # Add Student
    def add_student(self):

        try:
            roll_no = input("Enter Roll Number: ").strip()

            if not roll_no:
                raise ValueError("Roll number cannot be empty.")

            # Check duplicate roll number
            for student in self.students:
                if student.get_roll_no() == roll_no:
                    print("A student with this roll number already exists.")
                    return

            name = input("Enter Student Name: ").strip()

            if not name:
                raise ValueError("Name cannot be empty.")

            marks = float(input("Enter Marks: "))

            # Student constructor validates everything
            student = Student(roll_no, name, marks)

            self.students.append(student)

            print("Student added successfully!")

        except ValueError as error:
            print("Invalid input:", error)

    # Display Students
    def display_students(self):

        if not self.students:
            print("No student records found.")
            return

        print("\n========== STUDENT RECORDS ==========")

        for student in self.students:
            student.display_info()

    # Search Student
    def search_student(self):

        roll_no = input("Enter Roll Number to search: ").strip()

        for student in self.students:

            if student.get_roll_no() == roll_no:

                print("\nStudent Found!")
                student.display_info()
                return

        print("Student not found.")

    # Update Student
    def update_student(self):

        roll_no = input("Enter Roll Number to update: ").strip()

        for student in self.students:

            if student.get_roll_no() == roll_no:

                try:
                    new_name = input("Enter New Name: ").strip()

                    if not new_name:
                        raise ValueError("Name cannot be empty.")

                    new_marks = float(input("Enter New Marks: "))

                    # Test the new values first on a temporary student.
                    # If they are invalid this raises an error and the
                    # real record stays untouched.
                    Student(roll_no, new_name, new_marks)

                    student.set_name(new_name)
                    student.set_marks(new_marks)

                    print("Student record updated successfully!")
                    return

                except ValueError as error:
                    print("Invalid input:", error)
                    return

        print("Student not found.")

    # Delete Student
    def delete_student(self):

        roll_no = input("Enter Roll Number to delete: ").strip()

        for student in self.students:

            if student.get_roll_no() == roll_no:

                self.students.remove(student)

                print("Student deleted successfully!")
                return

        print("Student not found.")

    # Calculate Average Marks
    def calculate_average(self):

        if not self.students:
            print("No student records found.")
            return

        total = 0

        for student in self.students:
            total += student.get_marks()

        average = total / len(self.students)

        print("\nTotal Students :", len(self.students))
        print("Average Marks  :", round(average, 2))

    # Save Records
    def save_records(self):

        self.file_manager.save_students(self.students)


# ============================================================
# MAIN PROGRAM
# ============================================================

def main():

    manager = StudentManager()

    while True:

        print("\n======================================")
        print("       STUDENT MANAGEMENT SYSTEM")
        print("======================================")
        print("1. Add Student")
        print("2. Display Students")
        print("3. Search Student")
        print("4. Update Student")
        print("5. Delete Student")
        print("6. Calculate Average Marks")
        print("7. Save Records")
        print("8. Exit")
        print("======================================")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            manager.add_student()

        elif choice == "2":
            manager.display_students()

        elif choice == "3":
            manager.search_student()

        elif choice == "4":
            manager.update_student()

        elif choice == "5":
            manager.delete_student()

        elif choice == "6":
            manager.calculate_average()

        elif choice == "7":
            manager.save_records()

        elif choice == "8":
            manager.save_records()
            print("Thank you for using Student Management System!")
            break

        else:
            print("Invalid choice! Please enter 1-8.")


# Start Program
if __name__ == "__main__":
    main()
```
